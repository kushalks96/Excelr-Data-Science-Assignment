# Association RuleMining:

```python
unique_combinations = df['items'].tolist()
unique_combinations
```

```python
df
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
import pandas as pd
transactions = df['items'].apply(lambda x: x.split(',')).tolist()
re=TransactionEncoder()
re_array=re.fit(transactions).transform(transactions)
df1=pd.DataFrame(re_array,columns=re.columns_)
df1
```

```python
frequent_itemsets = apriori(df1, min_support=0.01, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.2)
rules
```

# Support itemsets

```python
from mlxtend.frequent_patterns import apriori
frequent_itemsets=apriori(df1,min_support=0.05,use_colnames=True)
frequent_itemsets
```

# Interview Questions:

1.What is lift and why is it important in Association rules?              
Ans:Lift is a measure used in association rule mining to evaluate the strength of an association rule compared to the expected likelihood of the items being associated if they were independent. In simpler terms, lift tells you how much more likely the items in the rule are to occur together than would be expected by chance.

2.What is support and Confidence. How do you calculate them?       
1.support          
Support measures how frequently an itemset appears in the dataset. It helps to identify how common or rare an itemset is.
Support of an itemset A is defined as the proportion of transactions in the dataset that contain the itemset A         
2.Confidence measures the likelihood that itemset B is purchased when itemset 𝐴 is purchased. It evaluates the reliability of the rule A→𝐵

3.What are some limitations or challenges of Association rules mining?

Ans:Some key limitations and challenges of Association Rules Mining include:

Scalability: Difficulty handling very large datasets due to exponential growth of itemsets.
Too Many Rules: Generates an overwhelming number of rules, many of which may be trivial or redundant.

Lack of Interestingness Measures: Standard metrics don't always capture truly novel or useful insights.

Sparse Data: Can struggle to find frequent itemsets in very sparse datasets.

Data Type Restrictions: Primarily designed for transactional data; handling continuous data requires extra steps.

Static Nature: Rules can become outdated quickly in dynamic environments.
