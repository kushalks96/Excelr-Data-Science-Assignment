Chi-Square Test

```python
import pandas as pd
from scipy.stats import chi2_contingency, chi
```

```python
data = {
    "Very Satisfied": [50, 70],
    "Satisfied": [80, 100],
    "Neutral": [60, 90],
    "Unsatisfied": [30, 50],
    "Very Unsatisfied": [20, 50]
}
```

```python
df = pd.DataFrame(data, index=["Smart Thermostat", "Smart Light"])
print("Observed Data:\n", df)
```

```python
chi2_stat, p_value, dof, expected = chi2_contingency(df)
```

```python
print("\nChi-Square Statistic:", chi2_stat)
print("Degrees of Freedom:", dof)
print("P-value:", p_value)
print("\nExpected Frequencies:\n", expected)
```

```python
alpha = 0.05
critical_value = chi.ppf(1 - alpha, dof)
print("\nCritical Value (0.05 significance):", critical_value)
```

```python
if chi2_stat > critical_value:
    print("\nReject H0: There is a significant association between device type and customer satisfaction.")
else:
    print("\nFail to Reject H0: No significant association between device type and customer satisfaction.")
```

Hypothesis Testing

```python
import math
from scipy.stats import norm
```

```python
sample_mean = 3050
X_mean = 600
sigma_X = 25
sigma = 5 * sigma_X
mu = 1000 + 5 * X_mean
n = 25
```

```python
z = (sample_mean - mu) / (sigma / math.sqrt(n))
print("Test Statistic (Z):", z)
```

```python
alpha = 0.05
critical_value = norm.ppf(1 - alpha)
print("Critical Value:", critical_value)
```

```python
if z > critical_value:
    print("Reject H0: There is evidence that the weekly operating cost is higher than the model predicts.")
else:
    print("Fail to Reject H0: There is no evidence that the weekly operating cost is higher than the model predicts.")
```
