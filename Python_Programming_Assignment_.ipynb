Prime Numbers

```python
import math

def is_prime(num):
  if num <= 1:
    return False
  for i in range(2, int(math.sqrt(num))+1):#does square root for num add 1 to maintain range
    if num % i == 0:
      return False
  return True
num=int(input("Enter a number: "))
if is_prime(num):
  print(num,'is a Prime Number')
else:
  print(num,'is not a Prime Number')
```

Product of Random Numbers

```python
import numpy as np
num1=np.random.randint(1,10)#generates random number
num2=np.random.randint(1,10)
correct_answer=num1*num2
user_answer=int(input(f"What is the product of {num1} and {num2}?"))
if user_answer==correct_answer:
  print("Correct!")
else:
  print(f"Incorrect. The correct answer is {correct_answer}")

```

Squares of Even Numbers

```python
# This script prints the squares of even numbers in the range of 100 to 200.
for num in range(100, 201):#set range for num b/w 100 to 200
  if num % 2 == 0:

    square = num ** 2 #squares numbers
    print(f"The square of {num} is {square}")
```

Word counter

```python
input_text = "This is a sample text. This text will be used to demonstrate the word counter."

words = [word.lower() for word in input_text.split()]#lower() conerts all words in lowercase,split() separates each words

word_count = {}#creates empty dic to assign word as ket and num as value


for w in words:
    if w in word_count:
        word_count[w] += 1
    else:
        word_count[w] = 1

print(word_count)
```

Check for Palindrome

```python
def is_palindrome(s):
    s = ''.join(c.lower() for c in s if c.isalnum())#c.isalnum() keeps only letters and numbers.
    return s == s[::-1] #checks if the cleaned string is the same backward.

print(is_palindrome("racecar"))
print(is_palindrome("madam"))
print(is_palindrome("Greetings"))
print(is_palindrome("Hello"))
print(is_palindrome("civic"))




```
